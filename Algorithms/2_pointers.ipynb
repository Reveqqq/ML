# Задача с собеседования в Яндекс

Написать функцию, которая меняет порядок слов в строке на противоположный, при этом не меняя расположение пробелов.

**Пример:**

```
__hello_my___dear_world_  →  __world_dear___my_hello_
```

- Слова разделены символом подчёркивания (`_`).
- Количество и расположение подчёркиваний между словами должно сохраниться.


## Асимптотика функции `change_word_order`

- Пусть $n$ — длина исходной строки.
- Разделение строки с помощью регулярного выражения работает за $O(n)$.
- Основной цикл с двумя указателями проходит по каждому элементу не более одного раза — это $O(n)$.
- Итоговая сборка строки — $O(n)$.

**Общая асимптотика:**

$$
O(n)
$$

Где $n$ — длина входной строки.

````python
import re

def change_word_order(s: str):
    s_split = [x for x in re.split(r'(_)', s) if x]
    l = 0
    r = len(s_split) - 1
    while l < r:
        s1 = s_split[l]
        s2 = s_split[r]
        if s1 != '_' and s2 != '_':
            s_split[l], s_split[r] = s_split[r], s_split[l] 
            l += 1
            r -= 1
        elif s1 != '_' and s2 == '_':
            r -=1
        elif s1 == '_' and s2 != '_':
            l += 1
        else:
            l += 1
            r -= 1

    return ''.join(s_split)
````

````python
input_s = '__hello_my___dear_world_'
new_word = change_word_order(input_s)
````

````python
print('before:', input_s)
print('after:', new_word)
````
